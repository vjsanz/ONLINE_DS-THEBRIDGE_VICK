## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

radianes = grados*(pi/180)

```python
grados = int(input( "grados = "))
pi = 3.14159265359
radianes = grados * (pi / 180)
print(f"{grados} grados son {radianes} radianes")
```

## Ejercicio 3
Escribe un programa que calcule el area de un paralelogramo (base x altura). También con `input`

```python
base = float(input("Introduce la base del paralelogramo: "))
altura = float(input("Introduce la altura del paralelogramo: "))
area = base * altura
print(f"El área del paralelogramo es: {area}")
```

## Ejercicio 4
Escribe un programa que calcule el area de un paralelogramo (base x altura). También con `input`. En este caso debe ser capaz de admitir valores reales, con decimales, para base y altura en español. (PISTA: Recuerda que tendrás que hacer uso del método replace)

```python
base = input("Introduce la base del paralelogramo: ")
base = base.replace(",", ".")
base = float(base)
altura = input("Introduce la altura del paralelogramo: ")
altura = altura.replace(",", ".")
altura = float(altura)
area = base * altura
print(f"El área del paralelogramo es: {area}")
```

## Ejercicio 5
Complete el siguiente código para solicitar al usuario su nombre y su edad. Luego, calcule en qué año nació el usuario y muestre el resultado.
